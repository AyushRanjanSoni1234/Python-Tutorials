**Question 5:** Write a Python program to calculate the factorial of a number.

```python
# Your code here
# 5! = 5*4*3*2*1
a = int(input("enter a number = "))
c = a
b = 1
while c > 0:
    b = b*c
    c = c-1 
print(f"Factorial of {a} = {b}")    
```

```python
def factorial(n):
    if n==1:
        return 1
    else:
        return n * factorial(n-1)
    
num = int(input("Enter a number :"))
print(f"Factorial of {num} = {factorial(num)}")    
```

**Question 19:** Write a Python program to check if a string is a palindrome.

```python
# Your code here

def palindrome_number(num):
    if isinstance(num, str):
        n = num[::-1]
        if n == num:
            return "Palindrome Number"
        else:
            return "not palindrome number"
    if isinstance(num, int):
        n = str(num)[::-1]
        if int(n) == num:
            return "Palindrome Number"
        else:
            return "not palindrome number"
        
a = int(input("enter a numberer :"))
print(f"the {a} is {palindrome_number(a)}")        
```

```python
# Email Varification....
import re  # Import the regular expression module

def validate_email(input_email):
    # Define the regular expression pattern for a valid email
    valid_regex = r"^[a-zA-Z0-9.!#$%&'*+/=?^_`{|}~-]+@[a-zA-Z0-9-]+(?:\.[a-zA-Z0-9-]+)*$"
    
    # Match the input email with the pattern
    if re.match(valid_regex, input_email):
        print("Valid email address!")
        return True
    else:
        print("Invalid email address!")
        return False

# Example usage
email = input("Enter an email address: ")
validate_email(email)
```
