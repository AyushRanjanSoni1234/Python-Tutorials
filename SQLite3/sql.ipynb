### SQLITE3

```python
import sqlite3
```

```python
## Connection to the database
conn = sqlite3.connect('example.db')
c = conn.cursor()

## Create table
c.execute('''
create table if not Exists users(
          id integer primary key autoincrement,
          name text,
          age integer,
          address text,
            phone text
          )
          ''')

## Insert a row of data

c.execute("insert into users (name, age, address, phone) values ('John Doe', 30, '123 Elm St', '555-1234')")
c.execute("insert into users (name, age, address, phone) values ('Jane Smith', 25, '456 Oak St', '555-5678')")  
c.execute("insert into users (name, age, address, phone) values ('Alice Johnson', 28, '789 Pine St', '555-8765')")
c.execute("insert into users (name, age, address, phone) values ('Bob Brown', 35, '321 Maple St', '555-4321')")
c.execute("insert into users (name, age, address, phone) values ('Charlie Black', 40, '654 Cedar St', '555-6789')")
c.execute("insert into users (name, age, address, phone) values ('Diana White', 22, '987 Birch St', '555-3456')")
c.execute("insert into users (name, age, address, phone) values ('Eve Green', 29, '159 Spruce St', '555-7890')")
c.execute("insert into users (name, age, address, phone) values ('Frank Blue', 32, '753 Fir St', '555-2345')")
c.execute("insert into users (name, age, address, phone) values ('Grace Yellow', 27, '852 Willow St', '555-6780')")
c.execute("insert into users (name, age, address, phone) values ('Hank Red', 38, '951 Chestnut St', '555-8901')")
c.execute("insert into users (name, age, address, phone) values ('Ivy Purple', 31, '357 Poplar St', '555-1235')")

## Save (commit) the changes
conn.commit()
```

```python
c.execute("SELECT * FROM users")
```

```python
c.execute("SELECT * FROM users WHERE age > 30")
```

```python
c.execute("SELECT * FROM users")
rows = c.fetchall()

for row in rows:
    print(row)  
```

```python
values = [('John Doe', 30, '123 Elm St', '555-1234'),
          ('ayush', 25, '456 Oak St', '555-5678'),
          ('sachin', 28, '789 Pine St', '555-8765')]

c.executemany('insert into users (name, age, address, phone) values (?, ?, ?, ?)', values
           )  # Using parameterized queries to prevent SQL injection

conn.commit()
```
