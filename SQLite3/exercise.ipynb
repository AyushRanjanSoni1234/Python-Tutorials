### Assignment 7: Handling Transactions

```python
## Write a Python function to insert multiple employees into the `employees` table in a single transaction. 
# Ensure that if any insertion fails, none of the insertions are committed.

def insert_multiple(cursor, conn, table_, values):
    try:
        cursor.executemany(f'''
        INSERT INTO {table_} (name, age, department) VALUES (?, ?, ?)''', (values))
    except sqlite3.Error as e:
        print(f"An error occurred: {e}")
    else:
        # If all insertions are successful, commit the transaction
        return conn.commit()

# Insert multiple employees in a single transaction
values = [('Alice', 30, 'HR'),
          ('Bob', 25, 'IT'),
          ('Charlie', 28, 'Finance'),
          ('David', 35, 'Marketing'),
          ('Eve', 22, 'Sales')]

insert_multiple(cursor, conn, 'employees', values)

fetch_all_employees(cursor, 'employees')
```

```python
## Write a Python function to update the age of multiple employees in a single transaction. 
# Ensure that if any update fails, none of the updates are committed.

def update_multiple_ages(cursor, conn, table_, updates):
    try:
        for id_, new_age in updates:
            cursor.execute(f'''
            UPDATE {table_} SET age = ? WHERE id = ?''', (new_age, id_))
    except sqlite3.Error as e:
        print(f"An error occurred: {e}")
    else:
        # If all updates are successful, commit the transaction
        return conn.commit()
    
# Update the age of multiple employees in a single transaction
# Assuming the IDs of the employees to be updated are 2, 3, and 4
updates = [(2, 26), (3, 29), (4, 36)]
update_multiple_ages(cursor, conn, 'employees', updates)
fetch_all_employees(cursor, 'employees')    
```

```python
## Modify the `employees` table to include a foreign key referencing the `id` column in the `departments` table.

def modify_employees_table_with_foreign_key(cursor, conn):
    # Step 1: Rename the existing table
    cursor.execute('''
        ALTER TABLE employees RENAME TO employees_old;
    ''')

    # Step 2: Create the new employees table with a foreign key
    cursor.execute('''
        CREATE TABLE employees (
            id INTEGER PRIMARY KEY AUTOINCREMENT,
            name TEXT NOT NULL,
            age INTEGER,
            department_id INTEGER,
            FOREIGN KEY (department_id) REFERENCES departments(id)
        );
    ''')

    # Step 3: Copy data from the old table into the new one
    # Note: Make sure the old table has the same columns except for department_id
    cursor.execute('''
        INSERT INTO employees (id, name, age)
        SELECT id, name, age FROM employees_old;
    ''')

    # Step 4: Drop the old table
    cursor.execute('''
        DROP TABLE employees_old;
    ''')

    conn.commit()
    return conn
```

```python
## Write a Python function to insert data into both the `departments` and `employees` tables, ensuring referential integrity.

def insert_department_and_employee(department_id, department_name, employee_id, name, age, department):
    conn = sqlite3.connect('test.db')
    cursor = conn.cursor()
    try:
        cursor.execute('''
            INSERT INTO departments (id, name)
            VALUES (?, ?)
        ''', (department_id, department_name))
        cursor.execute('''
            INSERT INTO employees (id, name, age, department, department_id)
            VALUES (?, ?, ?, ?, ?)
        ''', (employee_id, name, age, department, department_id))
        conn.commit()
        print("Department and employee inserted successfully.")
    except Exception as e:
        conn.rollback()
        print("Error occurred, transaction rolled back.")
        print(e)
    finally:
        conn.close()

# Test the function
insert_department_and_employee(1, 'Finance', 10, 'Zara', 28, 'Finance')
```
