### Assignment 3: Data Cleaning

```python
## Create a Pandas DataFrame with 3 columns and 5 rows filled with random integers. Introduce some NaN values. 
# Fill the NaN values with the mean of the respective columns.

data3 = pd.DataFrame(np.random.randint(1, 10, size=(5, 3)), columns=['A', 'B', 'C'])
data3.iloc[1, 1] = np.nan
data3.iloc[3, 2] = np.nan   
print("\nOriginal DataFrame with NaN values:")
print(data3)
data3.fillna(data3.mean(), inplace=True)
print("\nDataFrame after filling NaN values with column means:")
print(data3)
mean = data3.mean()
print("\nMean of each column:")
print(mean)
```

### Assignment 4: Data Aggregation

```python
## Create a Pandas DataFrame with 2 columns: 'Category' and 'Value'. Fill the 'Category' column 
# with random categories ('A', 'B', 'C') and the 'Value' column with random integers. 
# Group the DataFrame by 'Category' and compute the sum and mean of 'Value' for each category.

category = np.random.choice(['A', 'B', 'C'], size=10)
value = np.random.randint(1, 20, size=10)
df = pd.DataFrame({'category': category, 'value': value})
print("\nOriginal DataFrame:")
print(df)
grouped_value = df.groupby('category')['value'].agg(['sum', 'mean']).reset_index()
print("\nGrouped DataFrame with sum and mean of 'Value' for each category:")
print(grouped_value)
```

```python
## Create a Pandas DataFrame with 3 columns: 'Product', 'Category', and 'Sales'. 
# Fill the DataFrame with random data. Group the DataFrame by 'Category' and compute the total sales for each category.

products = np.random.choice(['Product1', 'Product2', 'Product3', 'Product4'], size=10)
category = np.random.choice(['A', 'B', 'C'], size=10)
sales = np.random.randint(100, 500, size=10)
df1 = pd.DataFrame({'Product': products, 'Category': category, 'Sales': sales})
print("\nOriginal DataFrame:")
print(df1)
grouped_sales = df1.groupby('Category')['Sales'].sum().reset_index()
print("\nGrouped DataFrame with total sales for each category:")
print(grouped_sales)

sales_report = df1.groupby(['Category', 'Product'])['Sales'].agg(['sum', 'mean']).reset_index()
print("\nSales report grouped by 'Category' and 'Product':")
print(sales_report)

df1.groupby(['Category', 'Product'])['Sales'].agg(['sum', 'mean']).unstack().fillna(0).plot(kind='bar', stacked=True)
plt.title('Sales Report')
```

```python
df1.groupby(['Category', 'Product'])['Sales'].sum().unstack().fillna(0).plot(kind='bar', stacked=True)
plt.title('Sales Report')
```

### Assignment 5: Merging DataFrames

```python
## Create two Pandas DataFrames with a common column. Merge the DataFrames using the common column.

df1 = pd.DataFrame({'Product': ['Product1', 'Product2', 'Product3'], 'Category': ['A', 'B', 'C']})
df2 = pd.DataFrame({'Product': ['Product1', 'Product2', 'Product4'], 'Sales': [100, 200, 300]})
print("\nDataFrame 1:")
print(df1)
print("\nDataFrame 2:")
print(df2)
merged_df = pd.merge(df1,df2, on ='Product', how='inner')
print("\nMerged DataFrame:")
print(merged_df)

merged_df.plot(kind='bar', x='Product', y='Sales')
plt.title('Product Wise Sales Report')
```

```python
sns.barplot(data=merged_df, x='Category', y='Sales', hue='Product')
plt.title('Categories Wise Sales Report')
plt.show()
```

### Assignment 6: Time Series Analysis

```python
## Create a Pandas DataFrame with a datetime index and one column filled with random integers. 
# Resample the DataFrame to compute the monthly mean of the values.

dates = pd.date_range('2023-01-01', periods=10, freq='D')
values = np.random.randint(1, 100, size=10)
df = pd.DataFrame({'Date': dates, 'Value': values})
df.set_index('Date', inplace=True)
print("\nOriginal DataFrame with datetime index:")
print(df)
monthly_mean = df.resample('ME').mean()
print("\nMonthly mean of values:")
print(monthly_mean)
```

```python
## Create a Pandas DataFrame with a datetime index ranging from '2021-01-01' to '2021-12-31' and one column filled with random integers. 
# Compute the rolling mean with a window of 7 days.

dates = pd.date_range('2021-01-01', '2021-12-31', freq='D')
values = np.random.randint(1, 100, size=len(dates))
df = pd.DataFrame({'Date': dates, 'Value': values})
df.set_index('Date', inplace=True)
print("\nOriginal DataFrame with datetime index:")
print(df)
rolling_mean = df.rolling(window=7).mean()
print("\nRolling mean with a window of 7 days:")
print(rolling_mean)
rolling_mean.plot(title='Rolling Mean with a window of 7 days')
plt.show()
```

### Assignment 7: MultiIndex DataFrame

```python
## Create a Pandas DataFrame with a MultiIndex (hierarchical index). 
# Perform some basic indexing and slicing operations on the MultiIndex DataFrame.

arrays = [
    ['A', 'A', 'B', 'B'],
    ['X', 'Y', 'X', 'Y']
]
tuples = list(zip(*arrays))
print(tuples)
index = pd.MultiIndex.from_tuples(tuples, names=['Category', 'Subcategory'])
data = np.random.randint(1, 100, size=(4, 3))
df = pd.DataFrame(data, index=index, columns=['Value1', 'Value2', 'Value3'])
print("\nOriginal MultiIndex DataFrame:")
print(df)
print("\nSlicing the MultiIndex DataFrame:")
print(df.loc['A'])
print("\nSlicing the MultiIndex DataFrame:")
print(df.loc[(slice('A', 'B'), slice(None)), :])
print("\nSlicing the MultiIndex DataFrame:")
print(df.loc[(slice(None), 'Y'), :])
```

```python
## Create a Pandas DataFrame with MultiIndex consisting of 'Category' and 'SubCategory'. 
# Fill the DataFrame with random data and compute the sum of values for each 'Category' and 'SubCategory'.

arrays =[
    ['A', 'A', 'B', 'B'],
    ['X', 'Y', 'X', 'Y']
]
data = np.random.randint(1, 100, size=(4, 3))
tuples = list(zip(*arrays))
index = pd.MultiIndex.from_tuples(tuples, names=['Category', 'Subcategory'])
df = pd.DataFrame(data, index=index, columns=['Value1', 'Value2', 'Value3'])
print("\nOriginal MultiIndex DataFrame:")
print(df)
grouped_sum = df.groupby(level=0).sum()
print("\nSum of values for each 'Category':")
print(grouped_sum)
grouped_sum = df.groupby(level=1).sum() 
print("\nSum of values for each 'SubCategory':")
print(grouped_sum)
grouped_sum = df.groupby(level=[0, 1]).sum()
print("\nSum of values for each 'Category' and 'SubCategory':")
print(grouped_sum)
grouped_sum = df.groupby(level=[0, 1]).sum().unstack()
print("\nSum of values for each 'Category' and 'SubCategory':")
print(grouped_sum)
grouped_sum = df.groupby(level=[0, 1]).sum().unstack().fillna(0)
grouped_sum.plot(kind='bar', stacked=True)
plt.title('Sum of values for each Category and SubCategory')
plt.show()
```

### Assignment 8: Pivot Tables

```python
## Create a Pandas DataFrame with columns 'Date', 'Category', and 'Value'. 
# Create a pivot table to compute the sum of 'Value' for each 'Category' by 'Date'.

dates = pd.date_range('2023-01-01', periods=10, freq='D')
category = np.random.choice(['A', 'B', 'C'], size=10)
values = np.random.randint(1, 100, size=10)
df = pd.DataFrame({'Date': dates, 'Category': category, 'Value': values})
print("\nOriginal DataFrame:")
print(df)
pivot_ = pd.pivot_table(df, values='Value', index='Date', columns='Category', aggfunc='sum').fillna(0).reset_index()
print("\nPivot Table:")
print(pivot_)
pivot_.plot(x='Date', kind='line', stacked=True)
plt.title('Pivot Table of Values by Date and Category')
plt.show()
```

```python
## Create a Pandas DataFrame with columns 'Year', 'Quarter', and 'Revenue'. 
# Create a pivot table to compute the mean 'Revenue' for each 'Quarter' by 'Year'.

years = np.random.choice([2021, 2022, 2023], size=10)
quarters = np.random.choice(['Q1', 'Q2', 'Q3', 'Q4'], size=10)
revenues = np.random.randint(1000, 5000, size=10)
df = pd.DataFrame({'Year': years, 'Quarter': quarters, 'Revenue': revenues})
print("\nOriginal DataFrame:")
print(df)
pivot_ = pd.pivot_table(df, values='Revenue', index='Year', columns='Quarter', aggfunc='mean').fillna(0).reset_index()
print("\nPivot Table:")     
print(pivot_)

# pivot_.plot(x='Year', kind='line', stacked=True)

plt.plot(pivot_['Year'], pivot_['Q1'], label='Q1')
plt.plot(pivot_['Year'], pivot_['Q2'], label='Q2')
plt.plot(pivot_['Year'], pivot_['Q4'], label='Q4')
plt.xlabel('Year')
plt.ylabel('Revenue')
plt.xscale('linear')
plt.xticks(pivot_['Year'])
plt.legend()
plt.title('Pivot Table of Revenue by Year and Quarter')
plt.show()
```

### Assignment 9: Applying Functions

```python
## Create a Pandas DataFrame with 3 columns and 5 rows filled with random integers. 
# Apply a function that doubles the values of the DataFrame.

df = pd.DataFrame(np.random.randint(1,10, size=(5,3)), columns=['A', 'B', 'C'])
print("\nOriginal DataFrame:")
print(df)
df = df.apply(lambda x:x*2)
print("\nDataFrame after applying function to double the values:")
print(df)
```

```python
## Create a Pandas DataFrame with 3 columns and 6 rows filled with random integers. 
# Apply a lambda function to create a new column that is the sum of the existing columns.

df = pd.DataFrame(np.random.randint(1,10, size=(6,3)), columns=['A', 'B', 'C'])
print("\nOriginal DataFrame:")
print(df)
df['D'] = df.apply(lambda x:x.sum(), axis=1)
print("\nDataFrame after applying lambda function to create new column 'D':")
print(df)
```
