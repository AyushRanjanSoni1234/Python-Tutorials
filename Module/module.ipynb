```python
## Data Serialization

import json

data = {
    'name': 'John',
    'age': 30,
    'city': 'New York'
}

json_string = json.dumps(data)
print(json_string)
print(type(json_string))

json_dict = json.loads(json_string)
print(json_dict)
print(type(json_dict))
```

```python
import csv

with open('data.csv', 'w', newline='') as file:
    writer = csv.writer(file)
    writer.writerow(['name', 'age', 'city'])
    writer.writerow(['John', 30, 'New York'])

with open('data.csv', 'r') as file:
    redaer = csv.reader(file)
    for row in redaer:
        print(row)    
```

```python
## Use the `math` module to calculate the greatest common divisor (GCD) of two numbers and the factorial of a number.

import math
gcd = math.gcd(12,18)
factorial = math.factorial(5)
print(f'Greatest common divisor: {gcd}')
print(f'Factorial: {factorial}')
```

```python
## Use the `datetime` module to print the current date, calculate the date 100 days from today, and determine the day of the week for a given date.

import datetime as dt
current_date = dt.date.today()
future_date = current_date + dt.timedelta(days=100)
day_of_week = current_date.strftime('%A')
future_day_of_week = future_date.strftime('%A')
print(f'Current date: {current_date}')
print(f'Date 100 days from today: {future_date}')
print(f'Day of the week: {day_of_week}')
print(f'Day of the week for future date: {future_day_of_week}')
```

```python
## Use the `random` module to generate a list of 5 random numbers between 1 and 50 and shuffle the elements of a list.

import random
random_numbers = [random.randint(1, 50) for _ in range(5)]
print(f'Random numbers: {random_numbers}')
# my_list = [1, 2, 3, 4, 5]
random.shuffle(random_numbers)
print(f'Shuffled list: {random_numbers}')
```
