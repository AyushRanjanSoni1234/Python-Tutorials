```python
## Write a function that writes a list of strings to a file named `output.txt`, with each string on a new line.

def write_file(file, lst):
    with open(file, 'w') as f:
        for item in lst:
            f.write(item + '\n')

    with open(file, 'r') as f:
        print(f.read())        
```

```python
## Write a function that copies the contents of a file named `source.txt` to a new file named `destination.txt`.

def copy_file(source, destination):
    with open(source, 'w') as f:
        f.write('This is a Source file.\n')
        f.write('This file is used to copy all content and paste it destination.txt file.\n')
    with open(source, 'r') as f:
        content = f.read() 

    with open(destination, 'w') as f:
        f.write(content)
```

```python
copy_file('source.txt', 'destination.txt')
```

```python
## Write a function that reads the contents of a file named `document.txt` and returns the number of words in the file.

def count_words(file):
    with open(file, 'r') as f:
        content = f.read()
        words = content.split()
        return len(words)
    
count_words('source.txt')    
```

```python
## Write a function that finds and replaces all occurrences of a given word in a file named `data.txt` with another word.

def replace_word(file, old_word, new_word):
    with open(file, 'r') as f:
        content = f.read()
        new_content = content.replace(old_word, new_word)
    with open(file, 'w') as f:
        f.write(new_content)    
```

```python
## Write a function that reads the contents of a file named `reverse.txt` and prints each line in reverse order.
def reverse_lines(file):
    content = []
    with open(file, 'r') as f:
        for line in f:
            content.append(line.strip()[::-1])

    with open('reverse.txt', 'w') as f:
        for line in content:
            f.write(line + '\n')        

reverse_lines('source.txt')            
```

```python
## Write a function that reads the contents of a file named `stats.txt` and returns the number of lines, words, and characters in the file.

with open('stats.txt', 'w') as f:
    f.write('This is a stats file.\n')
    f.write('This file is used to count number of lines, words and characters.\n')
    
```

```python
def count_line_word_char(file):
    with open('stats.txt', 'r') as f:
        content = f.read()
        words = content.split()
        lines = content.split('\n')
        return len(lines), len(words), len(content)
    
count_line_word_char('stats.txt')    
```

```python
## Write a function that merges the contents of multiple files into a single file named `merged.txt`.

def merge_files(files, destination):
    with open(destination, 'w') as f:
        for file in files:
            with open(file, 'r') as f1:
                content = f1.read()
                f.write(content + '\n')


merge_files(['source.txt', 'destination.txt', 'stats.txt'], 'merged.txt')
```

```python
## Write a function that splits a large file named `large.txt` into smaller files of 100 lines each.

def split_file(file):
    with open(file, 'r') as f:
        content = f.readlines()
        for i in range(0, len(content), 100):
            with open(f'{i}.txt', 'w') as f1:
                f1.writelines(content[i:i+100])

split_file('merged.txt')                
```

```python
def split_file(filename, lines_per_file):
    with open(filename, 'r') as file:
        lines = file.readlines()
    for i in range(0, len(lines), lines_per_file):
        with open(f'{filename}_part{i//lines_per_file + 1}.txt', 'w') as part_file:
            part_file.writelines(lines[i:i + lines_per_file])

# Test
split_file('source.txt', 100)
```

```python
## Write a function that creates a log file named `activity.log` and writes log messages with timestamps.

import datetime

def log_message(message, filename='activity.log'):
    timestamp = datetime.datetime.now().isoformat()
    with open(filename, 'a') as file:
        file.write(f'[{timestamp}] {message}\n')

# Test
log_message('This is a log message.')
```

```python
## Write a function that reads a binary file named `image.bin` and writes its contents to another binary file named `copy_image.bin`.

with open('image.bin', 'wb') as f:
    f.write(b'\x00\x01\x02\x03\x04\x05')

def copy_binary_file(source, destination):
    with open(source, 'rb') as f:
        content = f.read()
    with open(destination, 'wb') as f:
        f.write(content)    
```

```python
## Write a function that reads a CSV file named `data.csv` and prints its contents as a list of dictionaries.

import csv

with open('data.csv', 'w', newline='') as f:
    writer = csv.writer(f)
    writer.writerow(['Name', 'Age', 'City'])
    writer.writerow(['Alice', 24, 'New York'])
    writer.writerow(['Bob', 30, 'San Francisco'])
    writer.writerow(['Charlie', 45, 'Seattle'])

def read_csv(file):
    with open(file, 'r', newline='') as f:
        reader = csv.DictReader(f)
        for row in reader:
            print(dict(row))

read_csv('data.csv')                
```

```python
## Write a function that reads a JSON file named `data.json` and prints its contents as a Python dictionary.

import json

data = {
    'name': 'Alice',
    'age': 24,
    'city': 'New York'
}

with open('data.json', 'w') as f:
    json.dump(data, f)

def read_json(file):
    with open(file, 'r') as f:
        data = json.load(f)
        print(data)    

read_json('data.json')
```

```python
## Write a function that attempts to read a file named `protected.txt` and handles any permission errors gracefully by printing an error message.

with open('protected.txt', 'w') as f:
    f.write('This is a protected file.\n')

def read_file(file):
    try:
        with open(file, 'r') as f:
            print(f.read())
    except PermissionError:
        print('Permission denied to read the file.')

read_file('protected.txt')        
```

```python
def read_protected_file(filename):
    try:
        with open(filename, 'r') as file:
            print(file.read())
    except PermissionError as e:
        print(f"Permission error: {e}")

# Test
read_protected_file('protected.txt')
```
